```python
from pathlib import Path
from run_rdt import process_reaction_data
from molecule_view import build_reaction_view
```

```python
RXN_DIR = Path("reaction_intermediates/FBPA_h")
print(f"Reaction folder: {RXN_DIR}")
print(f"Exists: {RXN_DIR.exists()}")
```

## Low-level API: `process_reaction_data`

Returns a `ReactionProcessingResult` dataclass with per-molecule metadata. Pure computation, no file writes.

```python
result = process_reaction_data(RXN_DIR)
result
```

```python
print(f"Reaction:  {result.rxn_name}")
print(f"SMILES:    {result.smiles}")
print(f"Reactants: {result.from_num}")
print(f"Products:  {result.to_num}")
print(f"Molecules: {len(result.molecules)}")
print(f"Mapping:   {result.mapping_text[:80]}...")
```

```python
for mol in result.molecules:
    print(f"Mol {mol.mol_num}: {mol.species_id:12s} side={mol.side:4s} "
          f"atoms={len(mol.entries):2d}  inchi_order={mol.inchi_order}")
```

```python
gap = result.molecules[0]
for entry in gap.entries:
    print(f"  rdt={entry.rdt_atom_index:2d}  {entry.species_id:12s}  "
          f"{entry.side:4s}  {entry.element}#{entry.element_index}")
```

## High-level API: `build_reaction_view`

Returns a `ReactionView` with `MoleculeView` objects that support `_repr_html_` for rich Jupyter display and `__str__` for text output.

```python
view = build_reaction_view(RXN_DIR)
view
```

### Text rendering

```python
print(view)
```

```python
for mol in view.molecules:
    print(mol)
    print()
```

### Summary

```python
print(f"Summary: {view.summary}")
print(f"Mapping text length: {len(view.mapping_text)} chars")
print(f"Number of transition pairs: {view.mapping_text.count(',') + 1 if view.mapping_text else 0}")
```

### HTML rendering

In Jupyter, `display(view)` renders the rich HTML representation with styled tables. (Static notebook viewers will see the HTML output below.)

```python
from IPython.display import display
display(view)
```

```python
for mol in view.molecules:
    display(mol)
```
